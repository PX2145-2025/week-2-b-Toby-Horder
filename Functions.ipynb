## More about functions
Sometimes, there can be lots of variables that remain the same when calling a function. If that’s the case you can simplify the syntax a bit. For example, let’s write a function to find how long it takes a mass to hit the ground when dropped (from rest) from a height h. The function will look like:

````python
def T(h, g):
    '''
    Returns fall time [s] of a mass released from a height
    h [m] above the Earth's surface. g in ms^-2.
    '''
    return sqrt(2*h/g)
````

This is called as follows:

````python
height = 1  # metre
print ('Time from', height, 'metre', 'is' , T(height, 9.8),'s')
````

You *could* program this with a hard-coded value for g, to make the function call simpler:

````python
def T(h):
    '''
    Returns fall time [s] of a mass released from a height
    h [m] above the Earth's surface.
    '''
    g=9.81 # in ms^-2
    return sqrt(2*h/g)

height = 1 # metre
print ('Time from', height, 'metre', 'is' , T(height),'s')
````

but then of course it wouldn’t work for Mars’ gravity. A better way is

````python
def T(h, g=9.81):
    '''
    Returns fall time [s] of a mass released from a height
    h [m] above a planet's surface. Default g is 9.81 ms^-2
    '''
    return sqrt(2*h/g)

height = 1  # metre
print('Time from', height, 'metre', 'on Earth is' , T(height),'s')
print('Time from', height, 'metre', 'on Mars is' , T(height, g=3.7),'s')
````

````python
for height in [1,2,5,10]:
    print('Time from', height, 'metre', 'on Earth is' , T(height),'s')
````

A satellite has a circular orbit around the Earth with a period $T$.
  1. [2] Prove that its altitude $h$ above the Earth‘s surface is ($M$ is Earth’s mass and $R$ its radius):
$$h={\left ( \frac{GMT^2}{4\pi^2} \right )}^{1/3} - R$$

  3. [2] Write a program that requests user input of $T$ and then prints out the altitude in kilometre
  4. [2] Hence find the altitudes of satellites that orbit the Earth once a day (‘geosynchronous’ orbit), once every 90 minutes, and once every 45 minutes. What do you conclude from the last of these calculations?

$[1]$

$ m \omega^2 r = \frac{G m M}{r^2} \qquad $ [where m = mass of the satellite, M is the mass of the Earth, $\omega$ is the angular velocity and r is the distance from the center of the Earth to the satellite]



$ \omega^2 = \frac{G M }{r^3} \qquad $ [cancelling the masses and rearranging for omega on its own]


$ \omega = \frac{2 \pi}{T} \implies \omega^2 = \frac{4 \pi^2}{T^2} \qquad $ [where T is the time period of the satellite's orbit]

$ \frac{4 \pi^2}{T^2} = \frac{GM}{r^3}  \qquad $ [substituting in $\omega^2$]

$ r^3 = \frac{GM}{4 \pi^2} T^2 \qquad $ [rearranging for $r^3$]

$ r = h + R \qquad$ [where h = height above the surface of the earth and R = the radius of the Earth]


$ (h + R)^3 = \frac{GM}{4 \pi^2} T^2 \qquad $ [substituting in r = h + R]

$ h = (\frac{GM}{4 \pi^2} T^2)^\frac{1}{3} - R \qquad $ [rearranging for h]


````python
#2

import numpy as np

def orbitingHeightFromSurfaceOfEarth(TimePeriod):
  G = 6.67e-11 #m^3 kg^-1 s^-2
  M = 6e24 #kg
  R = 6400e3 #m
  h = ((G * M * TimePeriod**2)/(4 * np.pi**2))**(1/3) - R # h is the height of the satellite orbit, and the equation to calculate it
  return h #m

#3

areUnitsReadable = False

while areUnitsReadable == False: # a loop to make sure inputs are valid

  T = float(input("please enter the time period of the satellite: "))
  unitsForT = input("please enter the units for T from s,m,h,d,y for seconds,minutes,hours, days,years: ")

  #converting units if not in seconds
  if unitsForT == "s":
    T = T
    areUnitsReadable = True
  elif unitsForT == "m":
    T *= 60
    areUnitsReadable = True
  elif unitsForT == "h":
    T *= 60*60
    areUnitsReadable = True
  elif unitsForT == "d":
    T *= 60*60*24
    areUnitsReadable = True
  elif unitsForT == "y":
    T *= 60*60*24* 365
    areUnitsReadable = True
  else:
    print("units were not in a readable format, please try again")



h = orbitingHeightFromSurfaceOfEarth(T) #in m
h /= 1000 #convert to km


#4
print(f"the height of the satellite orbit is {h:.2f} km from the surface of the earth") #outputs height to 2 sig figs in km

#1 day outputs: the height of the satellite orbit is 35897.52 km from the surface of the earth
#90 minutes outputs: the height of the satellite orbit is 261.44 km from the surface of the earth
#45 minutes outputs: the height of the satellite orbit is -2203.55 km from the surface of the earth this is negative, which infers that for a satellite to orbit that quickly it would have to be inside the earth, therefore this isnt a possible height for a satellite
````

**IMPORTANT**: The `linfit` function itself must be self contained. It must only take in the x- and y-data, and return m, dm, c and dc. It must not do anything else, such as print or plot data - that should all be done outside the function.

````python
%%writefile LeastSquaresRegression.py

import numpy as np

def linfit(x,y): #x and y must be arrays or lists containing all of the x and y data

  x = np.array(x)
  y = np.array(y)

  pCoeff = np.polyfit( x, y, 1 )   #finding the polynomial which is a best fit to our data
  p = np.poly1d(pCoeff)

  #determining the mean of the data
  n = len(x)  #length of the array
  D = np.sum(x **2) - 1/n * np.sum(x)**2  #the independent coefficient

  pCoeff, residuals, _, _, _ = np.polyfit(x, y, 1, full=True)

  dmSquared = 1./(n-2)*residuals/D  #error in gradient squared
  dcSquared = 1./(n-2)*(D/n + np.mean(x)**2)*residuals/D #error in the y-intercept squared

  errorInGradient = dmSquared**(1/2)  #the error in the gradient
  errorInYintercept = dcSquared**(1/2)  #error in the yintercept
  gradient = pCoeff[0] #getting the gradient from the polynomial fit
  yIntercept = pCoeff[1] #getting the yintercept from the polynomial fit

  return errorInGradient, errorInYintercept, gradient, yIntercept, p, D, residuals
````
